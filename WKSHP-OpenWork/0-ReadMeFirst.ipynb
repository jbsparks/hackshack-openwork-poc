## Troubleshooting

If the OpenCode link doesn't load, run this cell to check the environment:

```bash
# Environment health check
echo "=== Tutorial Page ==="
STUDENT_NUM=$(echo $USER | grep -o '[0-9]*')
TUTORIAL_PORT=$((8080 + ${STUDENT_NUM:-0}))
OW_PORT=$((5178 + ${STUDENT_NUM:-0}))
if curl -s http://localhost:${TUTORIAL_PORT} > /dev/null 2>&1; then
    echo "[OK] Tutorial page running on port ${TUTORIAL_PORT}"
else
    echo "[ERROR] Tutorial page not responding on port ${TUTORIAL_PORT} -- contact a lab proctor"
fi

echo ""
echo "=== OpenCode ==="
if curl -s http://localhost:${OW_PORT} > /dev/null 2>&1; then
    echo "[OK] OpenCode web UI running on port ${OW_PORT}"
else
    echo "[WARN] OpenCode not responding on port ${OW_PORT} -- it may still be starting"
fi

echo ""
echo "=== Zen Models ==="
if curl -s --max-time 10 https://opencode.ai/zen/v1/models > /dev/null 2>&1; then
    echo "[OK] Zen API reachable"
else
    echo "[WARN] Cannot reach Zen API -- check network"
fi

echo ""
echo "=== Labs ==="
ls ~/labs/
```
